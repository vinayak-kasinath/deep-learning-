**Dataset: Iris**

```python
import numpy as np, pandas as pd, tensorflow as tf
from tensorflow.keras import layers, Sequential
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X, y = load_iris(return_X_y=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2, stratify=y, random_state=1)
sc = StandardScaler().fit(Xtr); Xtr, Xte = sc.transform(Xtr), sc.transform(Xte)

rows = []
for n_layers in (1, 2, 3):
    for act in ('relu', 'tanh', 'sigmoid'):
        m = Sequential([layers.Input((4,))] + [layers.Dense(16, activation=act) for _ in range(n_layers)]
                       + [layers.Dense(3, activation='softmax')])
        m.compile('adam', 'sparse_categorical_crossentropy', metrics=['accuracy'])
        m.fit(Xtr, ytr, epochs=100, batch_size=16, verbose=0)
        loss, acc = m.evaluate(Xte, yte, verbose=0)
        rows.append((n_layers, act, round(acc, 4), round(loss, 4)))
print(pd.DataFrame(rows, columns=['hidden_layers', 'activation', 'test_accuracy', 'test_loss']))
```
